# creación de un data set de ejemplo con valores nulos

acá tenemos un diccionario que contiene 4 columnas con 7 filas osea una tabla cada columna se representa como una lista y cada valor de la columna contiene un valor especifico para esos registros 

````python
data = {
    'edad': [25, 30, np.nan, 45, 50, np.nan,35],
    'salario': [30000, 45000, 50000, np.nan,60000, 55000, np.nan],
    'ciudad': ['madrid','barcelona',None, 'valencia', 'madrid','barcelona','valencia'],
    'experiencia_en_años': [2,5,7, np.nan, 10,8,4]
}
df = pd.DataFrame(data)
print("data set oringinal")
print(df)
print("\n"+"="*50+ "\n")
````

## extrategia de eliminación 
queremos eliminar datos de nuestro data set de pruebas como podemos hacerlo ??

### formas de eliminar informacion de un data set

- **A. eliminar filas con cualquier valor nulo** 
- **B. eliminar filas solo si TODAS las columnas son nulas**
- **C. eliminar filas con  nulos en columnas especificas**
- **D. eliminar columnas con valores nulos**
- **E. imputacion con valores constantes**

---

**NOTA**
cuando vayamos hacer operaciones como eliminacion e imputaciones es mejor hacerlas sobre una copia del data set original para proteger los datos iniciales


````python
## creamos copia del data frame original
df_copia = df.copy()
````

### A. eliminar filas con cualquier valor nulo


````python
# ✅ FORMA CORRECTA: dropna() ya devuelve una copia, no necesitas .copy() antes
df_sin_nulos = df_copia.dropna()

print("df sin nulos")
print(df_sin_nulos)
print(f"\nFilas originales: {len(df)}, Filas despues:{len(df_sin_nulos)}")
````

¡¡Perdimos mas de la mitad de los datos!!

 ### B. eliminar filas solo si TODAS las columnas son nulas


````python
# ✅ FORMA CORRECTA: dropna() ya devuelve una copia
df_todas_nulas = df_copia.dropna(how='all')
print(df_todas_nulas)
````

los datos se mantuvieron iguales por que no tenemos columnas con todos sus valores nulos

### C. eliminar filas con  nulos en columnas especificas


````python
# ✅ FORMA CORRECTA: dropna() ya devuelve una copia
df_columna_especifica = df_copia.dropna(subset=['edad'])
print("data frame eliminar informacion de columnas especificas")
print(df_columna_especifica)
````

### D. eliminar columnas con valores nulos


````python
# eliminar columnas que tengan al menos un valor nulo 
df_sin_columnas_nulas = df_copia.dropna(axis=1) # axis significa columnas
print("data set eliminando columnas con nulos")
print(df_sin_columnas_nulas)
````

se nos borro todo el data set por que en todas nuestras columnas tenemos al menos un valor nulo 

## tecnicas de imputación

la imputación es mas util por que no perdemos datos

### A. imputacion con la MEDIA (para datos numericos)


````python
### media a la edad
media_edad = df['edad'].mean()

## ✅ FORMA CORRECTA: Crear copia explícita antes de modificar
df_fillna = df_copia.copy()
df_fillna['edad'] = df_fillna['edad'].fillna(media_edad)
print(df_fillna)
````

### imputacion con la mediana ( mas robusta para outliers -> valores atipicos)

````python
### rellenar con la mediana
mediana_salario = df['salario'].median()
print(f"mediana del salario {mediana_salario}")
df_fillna['salario'] = df_fillna['salario'].fillna(mediana_salario)
print(df_fillna)
````

### imputacion con la moda (para datos categoricos)


````python
## rellenar con el valor mas frecuente
moda_ciudad = df['ciudad'].mode()[0] # mode devuelve una serie
df_fillna['ciudad'] = df_fillna['ciudad'].fillna(moda_ciudad)
print("\ndata set con ciudades imputadas")
print(df_fillna)
````

### E) Completando la imputación - experiencia_en_años

Ahora completamos nuestro DataFrame `df_fillna` imputando la columna que falta.
Usamos la **mediana** porque es más robusta ante valores atípicos (outliers).

````python
# Calculamos la mediana de experiencia desde el DataFrame ORIGINAL (df)
# Importante: siempre calcular estadísticas desde los datos originales, no desde datos ya imputados
mediana_experiencia = df['experiencia_en_años'].median()
print(f"Mediana de experiencia: {mediana_experiencia} años")

# Imputamos los valores nulos en experiencia_en_años
df_fillna['experiencia_en_años'] = df_fillna['experiencia_en_años'].fillna(mediana_experiencia)

print("\n✅ DataFrame completamente imputado:")
print(df_fillna)

# Verificamos que ya no hay valores nulos
print(f"\n🔍 Valores nulos restantes: {df_fillna.isnull().sum().sum()}")
````

### D) Imputación con un valor constante

````python
# ✅ FORMA CORRECTA: Crear copia nueva para esta técnica
df_fillna_constante = df_copia.copy()
df_fillna_constante['experiencia_en_años'] = df_fillna_constante['experiencia_en_años'].fillna(0)
print("experiencia con 0 donde habia nulos ")
print(df_fillna_constante)
````

## Técnicas Avanzadas de Imputación

Las técnicas anteriores (media, mediana, moda) son simples pero tienen una limitación:
**usan un único valor para toda la columna**, ignorando las relaciones entre variables.

Las técnicas avanzadas aprovechan la estructura de los datos para hacer imputaciones más inteligentes.

### F) Imputación por Grupos (Group-wise Imputation)

**¿Qué es?**
En lugar de usar la media/mediana de TODA la columna, usamos la media/mediana de un GRUPO específico.

**¿Cuándo usarla?**
Cuando existe una relación lógica entre variables. Por ejemplo:
- El salario puede variar según la ciudad
- La experiencia puede variar según el rango de edad

**Ventaja:** Produce valores más realistas porque respeta las diferencias entre grupos.

````python
# Creamos un nuevo dataset para demostrar la imputación por grupos
# Este dataset tiene más datos para que los grupos sean más representativos

data_grupos = {
    'ciudad': ['madrid', 'madrid', 'madrid', 'barcelona', 'barcelona', 'barcelona', 'valencia', 'valencia'],
    'salario': [35000, 40000, np.nan, 45000, 50000, np.nan, 30000, np.nan],
    'experiencia': [3, 5, 4, 6, 8, 7, 2, 3]
}
df_grupos = pd.DataFrame(data_grupos)

print("Dataset original con nulos:")
print(df_grupos)
print("\n" + "="*50)
````

````python
# Primero veamos la media de salario por ciudad
print("Media de salario por ciudad:")
print(df_grupos.groupby('ciudad')['salario'].mean())
print("\n" + "="*50)

# Ahora imputamos el salario usando la media de CADA ciudad
# transform() aplica una función a cada grupo y devuelve un resultado del mismo tamaño
df_grupos_imputado = df_grupos.copy()

df_grupos_imputado['salario'] = df_grupos.groupby('ciudad')['salario'].transform(
    lambda x: x.fillna(x.mean())  # Para cada grupo, rellenar nulos con la media del grupo
)

print("\n✅ Dataset con salario imputado por grupo (ciudad):")
print(df_grupos_imputado)
````

### Explicación paso a paso de groupby + transform:

```python
df.groupby('ciudad')['salario'].transform(lambda x: x.fillna(x.mean()))
```

1. **`groupby('ciudad')`** → Agrupa los datos por ciudad
2. **`['salario']`** → Selecciona solo la columna salario de cada grupo
3. **`transform()`** → Aplica una función a cada grupo y devuelve valores del mismo tamaño
4. **`lambda x: x.fillna(x.mean())`** → Para cada grupo:
   - `x` es el salario de ese grupo (ej: salarios de Madrid)
   - `x.mean()` calcula la media solo de ese grupo
   - `x.fillna(...)` rellena los nulos con esa media

**Resultado:** Cada ciudad tiene su propio valor de imputación basado en sus propios datos.

### G) KNN Imputer (K-Nearest Neighbors)

**¿Qué es?**
Imputa valores nulos basándose en los **K vecinos más cercanos** (registros más similares).
Es decir, busca las filas más parecidas y usa sus valores para rellenar el nulo.

**¿Cuándo usarla?**
- Cuando hay relaciones complejas entre variables
- Cuando las técnicas simples no capturan bien los patrones de los datos
- Para datos numéricos (requiere calcular distancias)

**Ventaja:** Considera TODAS las variables para encontrar similitudes, no solo una.

**Desventaja:** 
- Solo funciona con datos numéricos
- Puede ser lento con datasets muy grandes

````python
from sklearn.impute import KNNImputer

# Creamos un dataset numérico para KNN (solo columnas numéricas)
# KNN no funciona con datos categóricos directamente
data_knn = {
    'edad': [25, 30, np.nan, 45, 50, np.nan, 35, 28],
    'salario': [30000, 45000, 50000, np.nan, 60000, 55000, np.nan, 32000],
    'experiencia': [2, 5, 7, np.nan, 10, 8, 4, 3]
}
df_knn = pd.DataFrame(data_knn)

print("Dataset original para KNN:")
print(df_knn)
print("\nValores nulos:")
print(df_knn.isnull().sum())
````

````python
# Creamos el imputador KNN
# n_neighbors=2 significa que usará los 2 vecinos más cercanos para imputar
imputer = KNNImputer(n_neighbors=2)

# fit_transform() aprende los patrones y transforma los datos
# Devuelve un array de numpy, así que lo convertimos de vuelta a DataFrame
df_knn_imputado = pd.DataFrame(
    imputer.fit_transform(df_knn),  # Aplica KNN imputation
    columns=df_knn.columns          # Mantiene los nombres de columnas
)

print("✅ Dataset imputado con KNN (k=2):")
print(df_knn_imputado)
print("\n🔍 Valores nulos restantes:", df_knn_imputado.isnull().sum().sum())
````

### ¿Cómo funciona KNN Imputer?

**Ejemplo:** Para imputar la edad del registro 2 (que tiene NaN):

1. **Calcula distancias** entre el registro 2 y todos los demás usando las columnas que SÍ tiene valores (salario=50000, experiencia=7)

2. **Encuentra los K vecinos más cercanos** (los 2 más similares según sus valores)

3. **Calcula el promedio** de la edad de esos vecinos

4. **Imputa ese valor** en el registro 2

### ¿Cómo elegir el valor de K (n_neighbors)?

| Valor de K | Comportamiento |
|------------|----------------|
| K pequeño (1-3) | Más sensible a datos locales, puede ser ruidoso |
| K mediano (5-10) | Balance entre local y global |
| K grande (>10) | Más suave, tiende hacia la media general |

**Regla práctica:** Empieza con K = sqrt(n) donde n es el número de filas, o usa validación cruzada.

## Resumen: ¿Qué técnica usar?

| Técnica | Tipo de Datos | Cuándo Usar | Pros | Contras |
|---------|---------------|-------------|------|---------|
| **Media** | Numéricos | Datos simétricos sin outliers | Fácil, rápida | Sensible a outliers |
| **Mediana** | Numéricos | Datos con outliers o sesgados | Robusta a outliers | Ignora distribución |
| **Moda** | Categóricos | Variables categóricas | Única opción para categorías | Puede sobrerepresentar |
| **Constante** | Cualquiera | Cuando 0 o "desconocido" tiene sentido | Simple, explícita | Puede introducir sesgo |
| **Por Grupos** | Cualquiera | Relaciones entre variables | Más precisa que global | Necesita grupos representativos |
| **KNN** | Numéricos | Relaciones complejas | Considera múltiples variables | Lenta, solo numéricos |

### Flujo de decisión recomendado:

```
¿Es categórica? 
    → SÍ → Usar MODA (o imputación por grupos si hay relación con otra variable)
    → NO → ¿Tiene outliers?
              → SÍ → Usar MEDIANA
              → NO → ¿Hay relación con otras variables?
                        → SÍ → Usar Por Grupos o KNN
                        → NO → Usar MEDIA
```

## Resumen: Mejores prácticas con copias

### Regla de oro sobre copias en pandas:

#### ✅ **Con `dropna()` - NO necesitas `.copy()` antes**
```python
# ❌ INCORRECTO (crea 2 copias innecesarias):
df_limpio = df_copia.copy()
df_limpio = df_limpio.dropna()

# ✅ CORRECTO (1 sola copia):
df_limpio = df_copia.dropna()
```

**Razón:** `dropna()` **siempre devuelve una copia nueva** por defecto.

---

#### ✅ **Con `fillna()` + asignación a columna - SÍ necesitas `.copy()` antes**
```python
# ❌ INCORRECTO (modifica el DataFrame original):
df_limpio = df_copia  # ← Esto es una REFERENCIA, no una copia
df_limpio['edad'] = df_limpio['edad'].fillna(30)
# Ahora tanto df_limpio COMO df_copia están modificados

# ✅ CORRECTO (crea copia explícita):
df_limpio = df_copia.copy()
df_limpio['edad'] = df_limpio['edad'].fillna(30)
# Solo df_limpio está modificado, df_copia queda intacto
```

**Aclaración importante:** `fillna()` por sí solo SÍ devuelve una copia nueva.
El problema es cuando **asignamos a una columna** con `df['col'] = valor`, ahí estamos modificando el DataFrame directamente.

---

### Tabla resumen:

| Operación | ¿Necesitas `.copy()` antes? | ¿Por qué? |
|-----------|---------------------------|-----------|
| `dropna()` | ❌ NO | Ya devuelve una copia nueva |
| `df_nuevo = df.fillna(valor)` | ❌ NO | fillna() devuelve copia nueva |
| `df['col'] = df['col'].fillna(valor)` | ✅ SÍ | La asignación modifica el DataFrame |
| `df['col'] = valor` | ✅ SÍ | Asignación directa modifica el original |

### Regla simple para recordar:
> Si vas a usar `df['columna'] = algo`, primero haz `df = df.copy()`
